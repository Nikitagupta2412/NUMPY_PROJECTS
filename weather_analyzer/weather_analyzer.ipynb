```python
import numpy as np
```

```python
temp_data = np.array([
    [34, 36, 38, 35, 33, 37, 39],
    [30, 31, 32, 30, 29, 31, 32],
    [32, 33, 35, 34, 31, 33, 36]
])
```

```python
cities = ["Delhi" ,"Mumbai","Pune"]
```

```python
for city,row in zip(cities,temp_data):
  print(f"{city}:{row}")
```

```python
city_avg = np.mean(temp_data,axis=1)
print('the avg temp of the cities is :',city_avg)
```

```python
daily_avg = np.mean(temp_data,axis=0)
print("the avg temp of the cities across the week is :",daily_avg)
```

```python
highest_temp = np.max(temp_data)
print("the highest temp recorded is :",highest_temp ,"and the city is ",np.argmax(temp_data))
```

```python
lowest_temp = np.min(temp_data)
print('the lowest temp recorded is :',lowest_temp)
```

```python
city_highavg = np.argmax(city_avg)
print('The city with highest avg temp',cities[city_highavg],"with highest temp",city_highavg)
```

```python
extreme_heat_days = temp_data[temp_data > 35]
print("the days with highest temp recorded are", extreme_heat_days)
```

```python
high_temp_rec_each_city = np.max(temp_data,axis=1)
for city,column in zip(cities,high_temp_rec_each_city):
  print(f"highest temp recorded by each city :{city} = {column}")
```

```python
low_temp_rec_each_city = np.min(temp_data,axis=1)
for city,column in zip(cities,low_temp_rec_each_city):
  print(f"lowest temp recorded by each city :{city} = {column}")
```

```python
temp_above_35 = np.where(temp_data>35)
total_days = np.size(temp_above_35)
for city,rec in zip(cities,temp_above_35):
  print(f"No of Days on which above 35 temp recorded by each city :{city} = {rec}")
```

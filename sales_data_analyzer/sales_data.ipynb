```python
import numpy as np
```

```python
sales = np.array([
    [10, 12, 15],
    [20, 18, 22],
    [8, 9, 11],
    [14, 15, 17]
])
print(sales)
```

```python
print("dimension of the arrays",sales.ndim)
```

```python
print("shape of the arrays ",sales.shape)
```

```python
print("total sales across the stores",np.sum(sales,axis=1))
```

```python
print("total sales across the months ",np.sum(sales,axis=0))
```

```python
print("sales of the store 1",sales[0])
```

```python
print("sales of the second month",sales[:,1])
```

```python
print("sales of the store 2 in third month",sales[1,2])
```

```python
store_total = np.sum(sales,axis = 1 )
best_store = np.argmax(store_total)
month_total = np.sum(sales,axis = 0 )
maxmonth = np.argmax(month_total)
```

```python
print("best performing store", best_store)
print("best-performing month across all stores ", maxmonth)
```

```python
updated_sales = np.clip(sales*1.10,a_min=0,a_max=20)
updated_sales
```

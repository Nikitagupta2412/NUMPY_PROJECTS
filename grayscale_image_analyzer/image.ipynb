```python
import numpy as np
```

```python
image = np.array(
    [
        [45, 120, 200, 10],
        [80, 255, 150, 60],
        [30, 90, 180, 220],
        [0, 100, 210, 250],
    ]
)
```

```python
inverted_image = 255 - image
```

```python
bright_image = np.clip(image+50 , 0 , 255)
```

```python
print("original image\n" , image)
```

```python
print( "inverted image\n ",inverted_image)
```

```python
print("bright imsge\n",bright_image)
```

```python
black_and_white = np.where(image >128,255,0)
black_and_white
```

```python
center_crop = image[1:3,1:3]
center_crop
```

```python
dark_pixel_count = np.sum(image<100)
print("pixels with value less than 100",dark_pixel_count)
```

```python
bright_pixel_count = np.sum(image>200)
print("pixels with value more than 100",bright_pixel_count)
```

```python
avg_brightness = np.mean(image)
print("avg brightness of the image is",avg_brightness)
```

```python
dark_image = np.clip(image-50,0,255)
print("dark image\n",dark_image)
```

```python
rotate_image = np.rot90(image)
print(rotate_image)
print(np.rot90(image, 2))
```

```python
brightest_pixel = np.max(image)
print("Brightest pixel:", brightest_pixel)
```

```python
brightest_position = np.unravel_index(np.argmax(image), image.shape)

print("Brightest pixel:", brightest_pixel)
print("Position:", brightest_position)
```

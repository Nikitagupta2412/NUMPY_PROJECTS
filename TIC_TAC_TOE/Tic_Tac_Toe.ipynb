```python
import numpy as np
```

```python
board = np.zeros((3,3), dtype=int)
board
```

```python
#simulate a few moves
board[0,0] = 1
board
```

```python
board[1,1] = 1
board
```

```python
board[2,2] = 1
board
```

```python
board[0,1] = 2
board[1,0] = 2
print("current board state:\n" , board)
```

```python
#Checking winning status for player 1
main_diag = np.diag(board)
check= np.all(main_diag == 1)
print(check)
main_diag
```

```python
another_diag = np.diag(np.fliplr(board))
check = np.all(another_diag==1)
print(check)
another_diag
```

```python
all_rows = np.all(board == 1 , axis =1)
all_rows
```

```python
all_columns = np.all(board == 1 , axis = 0 )
all_columns
```

```python
Row_0 = np.all(board[0,:]== 1)
Row_0
```

```python
avai_moves = np.argwhere(board == 0)
avai_moves
```

```python
#Checking winning status for player 2
main_diag = np.diag(board)
main_diag
check = np.all(main_diag == 2)
check
```

```python
another_diag = np.diag(np.fliplr(board))
another_diag
check = np.all(another_diag==2)
check
```

```python
all_rows = np.all(board == 2,axis = 1)
all_rows
```

```python
all_columns = np.all(board == 2, axis =0)
all_columns
```

```python
#winner checking status
#player 1
p1_win = (
    np.all(main_diag==1)
    or np.all(another_diag==1)
    or np.any(np.all(board == 1, axis = 1))
    or np.any(np.all(board == 1 ,axis = 0))
)
```

```python
#player 2
p2_win = (
    np.all(main_diag == 2)
    or np.all(another_diag == 2)
    or np.any(np.all(board == 2, axis = 1))
    or np.any(np.all(board == 2 ,axis = 0))
)
```

```python
if p1_win:
  winner = "Player 1 (X)"
elif p2_win:
  winner = "Player 2 (0)"
elif len(avai_moves) == 0:
  winner = "DRAW"
else:
  winner ="In progress"
print("Winner",winner)
```

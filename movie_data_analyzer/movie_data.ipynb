```python
import numpy as np
```

```python
ratings = np.array([
    [4,7,8,10],
    [3,9,10,3],
    [3,9,10,10]
])
ratings
```

```python
ratings.shape
```

```python
avgratingofmovie = np.mean(ratings,axis=0)
avgratingofmovie
```

```python
highratingmovie = np.argmax(avgratingofmovie)
highratingmovie
```

```python
avgratingbyuser = np.mean(ratings,axis=1)
avgratingbyuser
```

```python
userhighrating = np.argmax(avgratingbyuser)
userhighrating
```

```python
highratedmovies = ratings >= 8
highratedmovies
ratingsofhighratedmovies = ratings[highratedmovies]
```

```python
print("ratings of the movies rated high ",ratingsofhighratedmovies)
```

```python
total = len(ratingsofhighratedmovies)
print("total no of high rated movies",total)
```

```python
minimumrating = np.min(ratings)
print("minimum rating acrooss the ratings ",minimumrating)
```

```python
maximumrating = np.max(ratings)
print("maximum rating across the ratings",maximumrating)
```

```python
normalizedratings = (ratings-minimumrating)/(maximumrating-minimumrating)
normalizedratings
print(np.round(normalizedratings,2))
```

```python
print("The highest rated movie is " , np.argmax(avgratingofmovie))
```

```python
import numpy as np
```

```python
students = np.array(["nikita","javid","yuvraj","kirti","sanik","tulip","kartik","owl"])
```

```python
student_data = np.array([
    [6,78,98],
    [2,65,99],
    [3,34,65],
    [7,34,98],
    [8,67,87],
    [9,23,76],
    [4,14,76],
    [10,76,98]
],dtype = float )
```

```python
weights = np.array([1.5,0.3,0.6])
```

```python
def predict_student_score():
  study_hours = int(input("enter the study hours"))
  attendance = int(input("enter attendance"))
  practice_test_score = int(input("enter the score of your practice test"))
  data = np.array([study_hours,attendance,practice_test_score])
  normalized_data = (data - (np.min(student_data ,axis = 0)))/((np.max(student_data , axis = 0)) - (np.min(student_data , axis = 0)))
  predicted_score = normalized_data @ weights
  final_score = np.clip(predicted_score , 0.0 , 100.0)
  print("final predicted score " , final_score)
```

```python
predict_student_score()
```

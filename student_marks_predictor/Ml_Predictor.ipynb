```python
import numpy as np
```

```python
students = np.array(["nikita","javid","yuvraj","kirti","sanik","tulip","kartik","owl"])
```

```python
student_data = np.array([
    [6,78,98],
    [2,65,99],
    [3,34,65],
    [7,34,98],
    [8,67,87],
    [9,23,76],
    [4,14,76],
    [10,76,98]
])
```

```python
actual_scores = np.array ([98,96,93,94,95,99,100,23])
```

```python
weights = np.array([1.5,0.3,0.6])
```

```python
column_min = np.min(student_data,axis = 0)
column_max = np.max(student_data,axis = 0)
normalized_data = (student_data - column_min)/(column_max-column_min)
print("The normalised data values are\n", normalized_data)
```

```python
raw_predicted_scores = student_data@weights
print("Raw predicted scores",raw_predicted_scores)
```

```python
normalized_predicted_scores = normalized_data@weights
print(" Normalized predicted scores are ",normalized_predicted_scores)
```

```python
Raw_error = actual_scores - raw_predicted_scores
print("Raw errors",Raw_error)
```

```python
normalized_Raw_error = actual_scores - normalized_predicted_scores
print("Normalized Raw errors",normalized_Raw_error)
```

```python
normalized_abs_error = np.abs(normalized_Raw_error)
print("NOrmalized absolute error" , normalized_abs_error)
```

```python
abs_error = np.abs(Raw_error)
print("Absolute eroors " ,abs_error)
```

```python
raw_avg_error = np.mean(abs_error)
print("avg error across the whole calculation ", raw_avg_error)
```

```python
normalized_avg_error = np.mean(normalized_abs_error)
print("normalized avg error across the whole calculation ", normalized_avg_error)
```

```python
best_indx = np.argmin(abs_error)
print("the closest prediction was",students[best_indx],"Actual score",actual_scores[best_indx],"Predicted score",raw_predicted_scores[best_indx],"Absolute error" ,abs_error[best_indx])
```

```python
normalised_best_indx = np.argmin(normalized_abs_error)
print("the closest prediction was",students[best_indx],"Actual score",actual_scores[best_indx],"Predicted score",normalized_predicted_scores[best_indx],"Absolute error" ,normalized_abs_error[best_indx])
```

```python
worst_indx = np.argmax(abs_error)
print("the farthest prediction was",students[worst_indx],"Actual score",actual_scores[worst_indx],"Predicted score",raw_predicted_scores[worst_indx],"Absolute error" ,abs_error[worst_indx])
```

```python
normalized_worst_indx = np.argmax(normalized_abs_error)
print("the farthest prediction was",students[normalized_worst_indx],"Actual score",actual_scores[normalized_worst_indx],"Predicted score",normalized_predicted_scores[normalized_worst_indx],"Absolute error" ,abs_error[normalized_worst_indx])
```

```python
categorize = np.where(normalized_predicted_scores>=80,"High",np.where(normalized_predicted_scores >= 60 , "average" , "low"))
for name,score,cat in zip (students,normalized_predicted_scores,categorize):
  print(f"{name}:{score:.2f}:{cat}")
```

```python
categorize = np.where(raw_predicted_scores>=80,"High",np.where(raw_predicted_scores >= 60 , "average" , "low"))
for name,score,cat in zip (students,raw_predicted_scores,categorize):
  print(f"{name}:{score:.2f}:{cat}")
```

```python
avg_actual_score = np.mean(actual_scores)
avg_raw_predicted_score = np.mean(raw_predicted_scores)
if avg_raw_predicted_score > avg_actual_score:
  print("THE MODEL IS HYPING A LOT ")
else:
  print("THE MODEL IS UNDERHYPING A LOT")
```

```python
avg_actual_score = np.mean(actual_scores)
avg_normalized_predicted_score = np.mean(normalized_predicted_scores)
if avg_normalized_predicted_score > avg_actual_score:
  print("THE MODEL IS HYPING A LOT ")
else:
  print("THE MODEL IS UNDERHYPING A LOT")
```

```python
print("the avg predicted scores are :",avg_raw_predicted_score)
print("the avg normalised predicted scores are :",avg_normalized_predicted_score)
print("the avg actual scores are :",avg_actual_score)
print("MAE",raw_avg_error)
```

```python
import numpy as np
```

```python
students = np.array(["Alice","Bob","Charlie","David","Eva"])
subjects = np.array(["python","maths","english","attendance"])
```

```python
data = np.array(
    [
        [85, 78, 92, 90],
        [55, 60, 58, 85],
        [95, 92, 98, 95],
        [40, 50, 45, 60],
        [70, 75, 80, 88],
    ]
)
```

```python
academic_avg = np.mean(data[:,0:3],axis=1)
print("The academic averages of the students are" , academic_avg)
```

```python
subject_avg = np.mean(data,axis = 0)
subject_avg
```

```python
top_performers = students[academic_avg>=80]
print("top performers of the class are ",top_performers)
```

```python
attendance= data[:,3]
attendance
```

```python
attendance_alert = students[attendance < 75]
print("students at high attendance alert",attendance_alert)
```

```python
sortedlist = students[np.argsort(academic_avg)]
rank = sortedlist[::-1]
print("rank of the students from highest to lowest ",rank)
```

```python
for i in range(len(rank)):
  print(f"Rank {i+1} : {rank[i]} with total marks obtained {academic_avg[i]} and total attendance {attendance[i]}")
```

```python
highest_subj_avg = subjects[np.argmax(subject_avg[0:3])]
print("subject with the highest avg value",highest_subj_avg)
```

```python
best_student = students[np.argmax(academic_avg)]
print("the brightest student is",best_student)
```

```python
perfect_students = students[np.where((academic_avg >= 80 ) & (attendance >= 75))]
print("These students are the one with perfect balance of attendance and studies",perfect_students)
```

```python
import numpy as np
```

```python
print("The predicted scores will be calculated on the basis of slope , study hours and intercept ")
```

```python
study_hours = np.array([2, 3, 5, 7, 8])
```

```python
actual_scores = np.array([50, 65, 75, 88, 95])
```

```python
stuhours_mean = np.mean(study_hours)
```

```python
actscores_mean = np.mean(actual_scores)
```

```python
#slope = sum((x - x_mean) * (y - y_mean))/((x - x_mean)**2)
slope = np.sum((study_hours - stuhours_mean) * (actual_scores - actscores_mean)) / np.sum((study_hours - stuhours_mean)**2)
slope
```

```python
#c = y_mean - m . x_mean
intercept = actscores_mean - slope * stuhours_mean
intercept
```

```python
predicted_score = (slope * study_hours) + intercept
predicted_score
```

```python
errors = predicted_score - actual_scores
errors
```

```python
Abserror = np.abs(errors)
MAE = np.mean(Abserror)
MAE
```

```python
new_hours = int (input("enter new score "))
predicted_new_score = (slope * new_hours) + intercept
print("Predicted new score will be :" , predicted_new_score)
```

```python
new_multiple_hours = np.array([3,4,5,6,7])
predicted_multiple_score = (slope * new_multiple_hours) + intercept
print("predicted_multiple_score will be " , predicted_multiple_score)
```

```python
best_prediction = np.argmin(Abserror)
print("best prediction is" , best_prediction ,"with study hours" ,study_hours[best_prediction],"with actual scores",actual_scores[best_prediction] ,"with the predicted scores",predicted_score[best_prediction] , "with a total error " , errors[best_prediction])
```

```python
worst_prediction = np.argmax(Abserror)
print("worst prediction is :" ,worst_prediction ,"with study hours" ,study_hours[worst_prediction],"with actual scores",actual_scores[worst_prediction] ,"with the predicted scores",predicted_score[worst_prediction] , "with a total error " , errors[worst_prediction])
```

```python
max_error = np.max(Abserror)
print("maximum error made by the model :" , max_error)
```

```python
mean_squared_error = np.mean((errors)**2)
print("mean squared error " , mean_squared_error)
```

```python
Root_mean_square_error = np.sqrt(mean_squared_error)
print("Root mean squared error will be :" , Root_mean_square_error)
```

```python
positive_error = (predicted_score>actual_scores)
print(positive_error)
```

```python
negative_error = (predicted_score<actual_scores)
print(negative_error)
```

```python
perfect_error = (predicted_score == actual_scores)
print(perfect_error)
```

```python
categories = np.where(Abserror <= 3 , "excellent",
                      np.where(Abserror <= 7 , "good" , "Needs improvement"))
```

```python
for i in range (len(study_hours)):
  print(f"Hours: {study_hours[i]} | Actual score: {actual_scores[i]} | Predicted score: {predicted_score[i]} | Category : {categories[i]}")
```

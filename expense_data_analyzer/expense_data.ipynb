```python
import numpy as np
```

```python
expense_data = np.array([
    [23,443,384],
    [73,94,934],
    [67,4903,4390],
    [783,4,43],
    [83,409,93],
    [3,442,23],
    [33,43,23]
])
expense_data
```

```python
total_spent = np.sum(expense_data)
total_spent
```

```python
category_totals = np.sum(expense_data,axis=0)
category_totals
```

```python
daily_totals = np.sum(expense_data,axis=1)
daily_totals
```

```python
expensive_days = np.where(daily_totals>= 500)
expensive_days
```

```python
print("total weekly expenditure",total_spent)
```

```python
print("total category wise spent " , category_totals)
```

```python
print("Daily total spent",daily_totals)
```

```python
print("days with spending high",expensive_days)
```

```python
highest_food_expense = np.max(expense_data[:,0])
highest_food_expense
```

```python
print("highest money spent on food on the day",highest_food_expense)
```

```python
avg_daily_transport = np.mean(expense_data[:,1])
avg_daily_transport
```

```python
print("avg money spent on transport",avg_daily_transport)
```

```python
costiest_day = np.argmax(daily_totals)
costiest_day
```

```python
print("the costiest day of the week ",costiest_day)
```

```python
cheapest_day = np.argmin(daily_totals)
cheapest_day
```

```python
exp_category = np.argmax(category_totals)
exp_category
```

```python
top_3_sorted = np.argsort(daily_totals)
top_3_expdays = top_3_sorted[-3:]
top_3_expdays
```
